# Bin Packing with CortexCloud

Pack items into the minimum number of capacity-limited bins using the free /v1/solvers/bin-packing preset.


```python
# Bin Packing with CortexCloud
#
# Pack items into the minimum number of capacity-limited bins using the free /v1/solvers/bin-packing preset.
#
# This notebook is executable on Colab — run all cells top to bottom.
# The first two cells are FREE (estimate + dry-run). The final cell pays
# $$0.05 USDC on Base via x402 when you set your private key (see cell 3).

import json
from cortexcloud import CortexCloud

cc = CortexCloud()

PROBLEM = {"problem_type": "qubo", "n": 18, "data": {"linear": [-0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5, -0.5], "quadratic": {"0,6": 1.0, "1,7": 1.0, "2,8": 1.0}}}

est = cc.estimate(PROBLEM)
print("recommended:", est["recommendation"]["mode"], "| solver:", est["recommendation"]["solver_id"],
      "| price: $", est["recommendation"]["cortexcloud_price_usd"],
      "| runtime:", est["recommendation"]["estimated_runtime_s"], "s")

sim = cc.simulate(PROBLEM)
print("feasible:", sim["feasible"], "| confidence:", sim["confidence"])
print("issues:", sim.get("issues", []))

# Optional: paid solve. Set your private key (wallet with USDC on Base).
# import os
# cc_paid = CortexCloud(private_key=os.environ["EVM_PRIVATE_KEY"])
# job = cc_paid.optimize(PROBLEM)
# print(cc_paid.wait(job["job_id"]))
```
